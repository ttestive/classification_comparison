## 1. Naive Bayes
### Introdução e Fundamentos
O algoritmo Naive Bayes é uma técnica de classificação baseada no Teorema de Bayes com a suposição de independência forte (naive) entre os preditores. É amplamente utilizado em tarefas como classificação de texto, filtragem de spam e análise de sentimento, devido à sua simplicidade, eficiência e bom desempenho em grandes conjuntos de dados.

### Aplicação e Exemplos
*   **Cenários de Uso**: Classificação de e-mails como spam ou não spam, categorização de documentos.
*   **Vantagens**: Rápido para treinar, bom com dados de alta dimensão, funciona bem com poucos dados de treinamento.
*   **Desvantagens**: A suposição de independência entre as features raramente é verdadeira na prática, o que pode afetar a precisão.

---

## 2. Árvore de Decisão
### Estrutura e Funcionamento
As Árvores de Decisão são modelos não-paramétricos que constroem um modelo de decisão na forma de uma estrutura de árvore. Elas dividem o conjunto de dados em subconjuntos menores com base em uma série de perguntas sobre as features, formando nós de decisão e folhas que representam as classes ou valores previstos. São intuitivas e fáceis de interpretar.

### Aplicação e Exemplos
*   **Cenários de Uso**: Diagnóstico médico, análise de risco de crédito, segmentação de clientes.
*   **Vantagens**: Simples de entender e visualizar, pode lidar com dados numéricos e categóricos, requer pouca preparação de dados.
*   **Desvantagens**: Tendência a overfitting, instabilidade (pequenas mudanças nos dados podem levar a uma árvore completamente diferente).

---

## 3. Random Forest
### Ensemble Learning com Árvores de Decisão
Random Forest é um algoritmo de aprendizado de máquina versátil que utiliza o conceito de ensemble learning, especificamente bagging, para combinar múltiplas árvores de decisão. Ele constrói um grande número de árvores de decisão durante o treinamento, cada uma treinada em uma amostra bootstrapped dos dados e considerando apenas um subconjunto aleatório de features para cada divisão. A previsão final é a moda (para classificação) ou a média (para regressão) das previsões das árvores individuais.

### Aplicação e Exemplos
*   **Cenários de Uso**: Previsão de preços de ações, detecção de fraude, previsão de doenças.
*   **Vantagens**: Reduz o overfitting em comparação com árvores de decisão únicas, robusto a ruídos e outliers, bom desempenho.
*   **Desvantagens**: Menos interpretável que uma única árvore de decisão, mais lento para treinar e fazer previsões que algoritmos mais simples.

---

## 4. Redes Neurais
### Fundamentos da Aprendizagem Profunda
As Redes Neurais, ou Redes Neurais Artificiais (RNAs), são modelos computacionais inspirados na estrutura e funcionamento do cérebro humano. Elas consistem em camadas interconectadas de nós (neurônios) que processam informações. São a base da aprendizagem profunda (Deep Learning) e são capazes de aprender padrões complexos em grandes volumes de dados, o que as torna poderosas para uma vasta gama de tarefas.

### Aplicação e Exemplos
*   **Cenários de Uso**: Reconhecimento de imagem, processamento de linguagem natural (NLP), sistemas de recomendação, jogos.
*   **Vantagens**: Capacidade de aprender padrões complexos e não-lineares, alto desempenho em tarefas complexas com grandes dados, flexibilidade.
*   **Desvantagens**: Requerem grandes volumes de dados para um bom treinamento, alto custo computacional, difícil interpretabilidade (caixa preta).

```r
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

dados = pd.read_csv(
    "dados/SMSSpamCollection",
    sep="\t",
    header=None,
    names=["label", "text"]
)

print(dados.head())
```

```r
#encoding
vectorizer = CountVectorizer(
    lowercase=True,
    stop_words="english"
)
```

```r

X = vectorizer.fit_transform(dados["text"])

y = dados["label"]
```

```r

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```r

previsoes = modelo.predict(X_teste)

# MATRIZ DE CONFUSÃO


matriz = confusion_matrix(
    y_teste,
    previsoes,
    labels=["ham", "spam"]
)

print(matriz)
```

# Random Forest

```r
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection          import train_test_split
from sklearn.ensemble                  import RandomForestClassifier
from sklearn.metrics                   import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

SEED = 71
```

```r
dados = pd.read_csv(
    "dados/SMSSpamCollection",
    sep="\t",
    header=None,
    names=["label", "text"]
)

print("Dimensões:", dados.shape)
print("\nDistribuição das classes:")
print(dados["label"].value_counts())
print("\nPrimeiras linhas:")
dados.head()
```

```r

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=5000,
    ngram_range=(1, 2)   # unigramas + bigramas
)

X = vectorizer.fit_transform(dados["text"])
y = dados["label"]

print(f"Matriz de features: {X.shape[0]} amostras × {X.shape[1]} termos")
```

```r

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y,
    test_size=0.20,
    random_state=SEED,
    stratify=y          # mantém proporção de classes
)

print(f"Treino : {X_treino.shape[0]} amostras")
print(f"Teste  : {X_teste.shape[0]} amostras")
```

```r

modelo = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    max_features="sqrt",   # considera √(nº features) por nó
    class_weight="balanced",
    random_state=SEED,
    n_jobs=-1
)

modelo.fit(X_treino, y_treino)
print("Floresta treinada ✓  |  árvores:", modelo.n_estimators)
```

```r

previsoes = modelo.predict(X_teste)

acuracia  = accuracy_score(y_teste, previsoes)
precisao  = precision_score(y_teste, previsoes, pos_label="spam")
recall    = recall_score(y_teste, previsoes, pos_label="spam")
f1        = f1_score(y_teste, previsoes, pos_label="spam")

print(f"Acurácia : {acuracia:.4f}")
print(f"Precisão : {precisao:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
print()
print(classification_report(y_teste, previsoes))
```

```r

matriz = confusion_matrix(y_teste, previsoes, labels=["ham", "spam"])

fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(
    matriz,
    annot=True, fmt="d",
    cmap="Greens",
    xticklabels=["ham", "spam"],
    yticklabels=["ham", "spam"],
    ax=ax
)
ax.set_xlabel("Previsto")
ax.set_ylabel("Real")
ax.set_title("Matriz de Confusão — Random Forest")
plt.tight_layout()
plt.show()
```

```r

importancias = pd.Series(
    modelo.feature_importances_,
    index=vectorizer.get_feature_names_out()
).nlargest(20)

fig, ax = plt.subplots(figsize=(7, 5))
importancias.sort_values().plot(kind="barh", ax=ax, color="#1D9E75")
ax.set_title("Top 20 termos mais importantes")
ax.set_xlabel("Importância (Gini)")
plt.tight_layout()
plt.show()
```

```r
dados = pd.read_csv(
    "dados/SMSSpamCollection",
    sep="\t",
    header=None,
    names=["label", "text"]
) #Lê os dados do nosso data set de SMS Spam, entendendo cada elemento com label("classe") e texto

print(dados.head())
```

```r
vectorizer = CountVectorizer(
    lowercase=True,
    stop_words="english"
) # Tudo minúsculo e remove palavras desnecessárias
# É algo padrão da biblioteca CountVextorizer que remove palavras da lingua inglesa, como The, are...
```

```r
X = vectorizer.fit_transform(dados["text"]) # Features (palavras)

y = dados["label"] # Target (ham ou spam)
```

```r
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.20, # 20% para teste
    random_state=71 #SEED para reprodução
)
```

```r
modelo = MLPClassifier(
    hidden_layer_sizes=(100, 50),  # 2 camadas de processamento
    #100 e 50 neuronios
    activation="relu", # Função de ativação
    max_iter=300, #Faz no máximo 300 ciclos de treino
    random_state=71 #SEED de reprodução
)

modelo.fit(X_treino, y_treino)
```

```r
previsoes = modelo.predict(X_teste)

matriz = confusion_matrix( #Matriz de confusão
    y_teste,
    previsoes,
    labels=["ham", "spam"]
)

print(matriz)
```

```r
#Aqui acontece o calculo das metricas, usando os dados de teste,
#definidos com o valor de Y
acuracia = accuracy_score(y_teste, previsoes)
precisao = precision_score(
    y_teste,
    previsoes,
    pos_label="spam"
)

recall = recall_score(
    y_teste,
    previsoes,
    pos_label="spam"
)

f1 = f1_score(
    y_teste,
    previsoes,
    pos_label="spam"
)
```

```r
print("Acurácia:", acuracia)
print("Precisão:", precisao)
print("Recall:", recall)
print("F1 Score:", f1)
```

```r
library(tm)
library(e1071)
library(caret)
```

```r
dados <- read.csv(
  "dados/SMSSpamCollection",
  sep = "\t",
  header = FALSE,
  col.names = c("label", "text"),
  stringsAsFactors = FALSE
)

head(dados)
```

```r
corpus <- Corpus(VectorSource(dados$text))

corpus <- tm_map(corpus, content_transformer(tolower))
corpus <- tm_map(corpus, removeWords, stopwords("english"))
corpus <- tm_map(corpus, removePunctuation)
corpus <- tm_map(corpus, stripWhitespace)

dtm <- DocumentTermMatrix(corpus)
```

```r
X <- as.data.frame(as.matrix(dtm))

y <- as.factor(dados$label)
```

```r
set.seed(42)

indice <- createDataPartition(y, p = 0.80, list = FALSE)

X_treino <- X[indice, ]
X_teste  <- X[-indice, ]
y_treino <- y[indice]
y_teste  <- y[-indice]
```

```r
library(nnet)

X_treino_matrix <- as.matrix(X_treino)
X_teste_matrix  <- as.matrix(X_teste)

modelo <- nnet(
  x          = X_treino_matrix,
  y          = class.ind(y_treino),
  size       = 50,        # neurônios na camada oculta
  maxit      = 300,       # iterações máximas
  decay      = 1e-4,      # regularização (equivalente ao weight decay)
  softmax    = TRUE,      # classificação multiclasse
  trace      = FALSE
)
```

```r
previsoes_prob <- predict(modelo, X_teste_matrix, type = "raw")

previsoes <- factor(
  colnames(previsoes_prob)[apply(previsoes_prob, 1, which.max)],
  levels = levels(y_teste)
)

matriz <- confusionMatrix(
  previsoes,
  y_teste,
  positive = "spam"
)

print(matriz$table)
```

```r
acuracia <- matriz$overall["Accuracy"]

precisao <- matriz$byClass["Precision"]

recall   <- matriz$byClass["Recall"]

f1       <- matriz$byClass["F1"]
```
